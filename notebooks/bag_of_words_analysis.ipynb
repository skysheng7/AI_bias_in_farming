[2] Exploratory : Second list of interests (refined based on the first)
- I'll put all the images that contains the word of interest in its GPT-4o description text into a seperate folder,
AFTER deleting those images that contains words in the negative list.
- For example, all images that contain "grazing", but does not contain ANY of the words in dairy_extensive_negative_list, will be placed into "results/cluster_eda2/dairy/extensive/grazing"

```python
# dairy 
dairy_extensive = ["grazing", "graze", "pasture", "pastures"]
dairy_extensive_negative_list = ["ceiling", "metal railing", "metal railings", "barn spacious", "spacious barn", 
                                 "bars", "aisle", "illumination", "interior large", "large lit", "large indoor", 
                                 "spacious rows", "metal pens", "beams high", "barn rows", "barn lit"]
dairy_intensive = ["ceiling", "metal railing", "metal railings", "barn spacious", "spacious barn", 
                   "bars", "aisle", "illumination", "interior large", "large lit", "large indoor", "spacious rows", 
                   "metal pens", "beams high", "barn rows", "barn lit"]
dairy_intensive_negative_list = ["grazing", "graze", "pasture", "pastures"]

# pig
pig_extensive = ["grazing", "graze", "pasture", "pastures", "grass", "soil", "green", "mud", "muddy", 
                 "rural"]
pig_extensive_negative_list = ["ceiling", "ceilings", "metal railing", "metal railings", "barn spacious", "spacious barn", 
                               "bars", "aisle", "illumination", "interior large", "large lit", "indoor", 
                               "spacious rows", "metal pens", "beams high", "lighting bright", "barn rows", 
                               "narrow barn", "industrial", "barn lit", "concrete", "central walkway", 
                               "light streaming", "interior barn", "rows pens", "barn floor", "closely packed"]
pig_intensive = ["floor", "ceiling", "metal railing", "metal railings", "barn spacious", 
                 "spacious barn", "bars", "aisle", "illumination", "large lit", 
                 "indoor", "spacious rows", "metal pens", "typical modern", "barn lit", 
                 "concrete", "central walkway"]
pig_intensive_negative_list = ["grazing", "graze", "pasture", "pastures", "grass", "soil", "outdoor", 
                               "pastoral", "scattered landscape", "vegetation"]

```

- Warning: the code cell below takes a long time to run (~ 30 minutes), please don't run again, as it's time and resource consuming.

```python
# Set up base paths
model = "dall-e-3"
source_base = Path("..") / "results"
dest_base = Path("..") / "results" / "cluster_eda3"

# Process dairy extensive
module6_extract_pic_feature_words.process_farm_images(
    gpt4o_description_1_2gram,
    "dairy",
    "extensive",
    dairy_extensive,
    source_base,
    dest_base,
    model,
    grouping_style="combine",
    negative_word_list= dairy_extensive_negative_list
)

# Process dairy intensive
module6_extract_pic_feature_words.process_farm_images(
    gpt4o_description_1_2gram,
    "dairy",
    "intensive",
    dairy_intensive,
    source_base,
    dest_base,
    model,
    grouping_style="combine",
    negative_word_list= dairy_intensive_negative_list
)

# Process pig extensive
module6_extract_pic_feature_words.process_farm_images(
    gpt4o_description_1_2gram,
    "pig",
    "extensive",
    pig_extensive,
    source_base,
    dest_base,
    model,
    grouping_style="combine",
    negative_word_list = pig_extensive_negative_list
)

# Process pig intensive
module6_extract_pic_feature_words.process_farm_images(
    gpt4o_description_1_2gram,
    "pig",
    "intensive",
    pig_intensive,
    source_base,
    dest_base,
    model,
    grouping_style="combine",
    negative_word_list = pig_intensive_negative_list
)
```
